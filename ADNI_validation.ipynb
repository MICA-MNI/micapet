```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nilearn
import brainstat
import brainspace
from nilearn import surface
from brainstat.datasets import fetch_mask, fetch_template_surface
from brainspace.plotting import plot_hemispheres
from glob import glob
import os
```

```python
derivatives_MK6240="/host/verges/tank/data/jack/MK6240/BIDS_test_MK6240_ADNI4/derivatives/MICA-PET_v.alpha.0.5"

pial_left, pial_right = fetch_template_surface("fslr32k",join=False)
mask = fetch_mask("fslr32k")
T_MK6240 = pd.read_csv("/host/verges/tank/data/jack/MK6240/BIDS_test_MK6240_ADNI4/tmp/T1w_4_07_2026.csv")
```

```python
T_CN_MK6240 = T_MK6240[T_MK6240.Group == "CN"]
T_MCI_MK6240 = T_MK6240[T_MK6240.Group == "MCI"]
T_AD_MK6240 = T_MK6240[T_MK6240.Group == "AD"]
```

```python
MK6240_surf_CN_L = []
MK6240_surf_CN_R = []
for subj in T_CN_MK6240.Subject_short:
    MK6240_surf_CN_L.append(nilearn.surface.load_surf_data(os.path.join(derivatives_MK6240, f"sub-{subj}", "ses-01", "surf", f"sub-{subj}_ses-01_hemi-L_surf-fsLR-32k_label-midthickness_pvc-none_ref-cerebellarGM_smooth-10mm_trc-MK6240_pet.func.gii")))
    
    MK6240_surf_CN_R.append(nilearn.surface.load_surf_data(os.path.join(derivatives_MK6240, f"sub-{subj}", "ses-01", "surf", f"sub-{subj}_ses-01_hemi-R_surf-fsLR-32k_label-midthickness_pvc-none_ref-cerebellarGM_smooth-10mm_trc-MK6240_pet.func.gii")))

to_plot = np.hstack((np.mean(MK6240_surf_CN_L, axis=0), np.mean(MK6240_surf_CN_R, axis=0)))
to_plot[~mask] = np.nan

plot_hemispheres(pial_left, pial_right,to_plot,color_bar=True,embed_nb=True,transparent_bg=False,zoom=1.25,color_range=(0,1.5),size=(900,200),nan_color=(0.7, 0.7, 0.7, 1))
```

```python
MK6240_surf_MCI_L = []
MK6240_surf_MCI_R = []
for subj in T_MCI_MK6240.Subject_short:
    MK6240_surf_MCI_L.append(nilearn.surface.load_surf_data(os.path.join(derivatives_MK6240, f"sub-{subj}", "ses-01", "surf", f"sub-{subj}_ses-01_hemi-L_surf-fsLR-32k_label-midthickness_pvc-IY_ref-cerebellarGM_smooth-10mm_trc-MK6240_pet.func.gii")))
    
    MK6240_surf_MCI_R.append(nilearn.surface.load_surf_data(os.path.join(derivatives_MK6240, f"sub-{subj}", "ses-01", "surf", f"sub-{subj}_ses-01_hemi-R_surf-fsLR-32k_label-midthickness_pvc-IY_ref-cerebellarGM_smooth-10mm_trc-MK6240_pet.func.gii")))

to_plot = np.hstack((np.mean(MK6240_surf_MCI_L, axis=0), np.mean(MK6240_surf_MCI_R, axis=0)))
to_plot[~mask] = np.nan

plot_hemispheres(pial_left, pial_right,to_plot,color_bar=True,embed_nb=True,transparent_bg=False,zoom=1.25,color_range=(0,1.5),size=(900,200),nan_color=(0.7, 0.7, 0.7, 1))
```
